# FR-MF - Rainfall rate over France mainland

This notebook describes how to access and use the zarr-version of the French radar data provided by [Météo-France](https://donneespubliques.meteofrance.fr/). 

The dataset includes aproximately ~500.000 5 min timesteps covering a 1536 x 1536 resolution with 1km grid-size. The original .npz files are stored in (mm x 10^-2) and the .zarr dataset is in precipitation rate (mm/h).

The `v0.1.0` zarr dataset covers the temporal period of `2020-01-01` to `2024-12-31`.

```python
import matplotlib.pyplot as plt
import xarray as xr

import cartopy.feature as cfeature
import cartopy.crs as ccrs
import mlcast_datasets
```

```python
cat = mlcast_datasets.open_catalog()
list(cat.precipitation)
```

# FR-MF - 5-min Rainfall rate

The French MF SRI dataset is available in the intake catalog as [name_ds_on_catalog]()

```python
ds = cat.fr_mf_prate_5min.to_dask()
```

```python
var_name = "prate"
crs_name = ds[var_name].grid_mapping
data_crs = ccrs.Projection(ds[crs_name].crs_wkt)

g = (
    ds[var_name]
    .sel(time="2023-11-17")
    .isel(time=slice(None, 3))
    .plot(
        transform=data_crs,
        cmap="viridis",
        add_colorbar=True,
        col="time",
        robust=True,
        subplot_kws=dict(projection=data_crs),
        vmin=0,
    )
)

for ax in g.axs.flat:
    ax.coastlines()
    ax.gridlines()
```
